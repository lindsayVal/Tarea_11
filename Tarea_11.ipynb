```python
import numpy as np
import logging

# Configuración de logging para ver el progreso de las iteraciones
logging.basicConfig(level=logging.INFO, format='%(message)s')

def gauss_jacobi(*, A: np.ndarray, b: np.ndarray, x: np.ndarray, tol: float, max_iter: int):
    """Resuelve el sistema mediante el método de Jacobi."""
    if not isinstance(A, np.ndarray): A = np.array(A, dtype=float)
    if not isinstance(b, np.ndarray): b = np.array(b, dtype=float)
    if not isinstance(x, np.ndarray): x = np.array(x, dtype=float, ndmin=2).T
    
    n = A.shape[0]
    trayectoria = [x.copy()]

    for k in range(1, max_iter + 1):
        x_new = np.zeros((n, 1))
        for i in range(n):
            suma = sum([A[i, j] * x[j] for j in range(n) if j != i])
            x_new[i] = (b[i] - suma) / A[i, i]

        trayectoria.append(x_new.copy())
        if np.linalg.norm(x_new - x) < tol:
            return x_new, trayectoria
        x = x_new.copy()
        logging.info(f"Jacobi Iter {k}: {x.T}")
    return x, trayectoria

def gauss_seidel(*, A: np.ndarray, b: np.ndarray, x: np.ndarray, tol: float, max_iter: int):
    """Resuelve el sistema mediante el método de Gauss-Seidel[cite: 16]."""
    if not isinstance(A, np.ndarray): A = np.array(A, dtype=float)
    if not isinstance(b, np.ndarray): b = np.array(b, dtype=float)
    if not isinstance(x, np.ndarray): x = np.array(x, dtype=float, ndmin=2).T
    
    n = A.shape[0]
    trayectoria = [x.copy()]

    for k in range(1, max_iter + 1):
        x_new = np.zeros((n, 1))
        for i in range(n):
            suma = sum([A[i, j] * x_new[j] for j in range(i)]) + \
                   sum([A[i, j] * x[j] for j in range(i + 1, n)])
            x_new[i] = (b[i] - suma) / A[i, i]

        trayectoria.append(x_new.copy())
        if np.linalg.norm(x_new - x) < tol:
            return x_new, trayectoria
        x = x_new.copy()
        logging.info(f"Gauss-Seidel Iter {k}: {x.T}")
    return x, trayectoria

# --- DEFINICIÓN DE SISTEMAS DEL EJERCICIO 1 ---

sistemas = {
    "1.a": {
        "A": [[3, -1, 1], [3, 6, 2], [3, 3, 7]], # [cite: 3, 8, 9]
        "b": [1, 0, 4] # [cite: 3, 8, 9]
    },
    "1.b": {
        "A": [[10, -1, 0], [-1, 10, -2], [0, -2, 10]], # [cite: 5, 17, 18]
        "b": [9, 7, 6] # [cite: 7, 17, 18]
    },
    "1.c": {
        "A": [[10, 5, 0, 0], [5, 10, -4, 0], [0, -4, 8, -1], [0, 0, -1, 5]], # [cite: 10, 13, 14, 15]
        "b": [6, 25, -11, -11] # [cite: 11, 13, 14, 15]
    },
    "1.d": {
        "A": [[4, 1, 1, 0, 1], [-1, -3, 1, 1, 0], [2, 1, 5, -1, -1], [-1, -1, -1, 4, 0], [0, 2, -1, 1, 4]], # [cite: 20, 21, 22]
        "b": [6, 6, 6, 6, 6] # [cite: 19, 20, 21, 22]
    }
}

# --- EJECUCIÓN ---

for clave, data in sistemas.items():
    print(f"\n{'='*10} {clave} {'='*10}")
    A_mat = np.array(data["A"])
    b_vec = np.array(data["b"])
    n_vars = A_mat.shape[0]
    
    print(f"\n--- MÉTODO DE JACOBI (2 iteraciones, x0=0) ---") # 
    sol_j, _ = gauss_jacobi(A=A_mat, b=b_vec, x=np.zeros(n_vars), tol=1e-3, max_iter=2)
    
    print(f"\n--- MÉTODO DE GAUSS-SEIDEL (2 iteraciones, x0=0) ---") # [cite: 16]
    sol_gs, _ = gauss_seidel(A=A_mat, b=b_vec, x=np.zeros(n_vars), tol=1e-3, max_iter=2)
```
